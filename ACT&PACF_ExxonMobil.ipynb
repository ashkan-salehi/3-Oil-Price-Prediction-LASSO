```python
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.graphics.tsaplots import plot_pacf
from statsmodels.tsa.stattools import pacf
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.stattools import acf


# close price for Exxon Mobil in 2025
df=pd.read_csv("STOCK_US_XNYS_XOM.csv")
df.head()

sliced_df = {}

df["Date"] = pd.to_datetime(df["Date"])

#tarikh = ["2025-05-20", "2025-06-20", "2025-07-20", "2025-08-20", "2025-09-20"]
#tarikh = ["2025-05-20", "2025-06-20", "2025-07-20"]
#tarikh = ["2025-12-20", "2025-12-21", "2025-12-22", "2025-12-23", "2025-12-24"
#,"2025-12-25"]
tarikh = ["2025-12-20", "2025-11-21", "2025-10-22", "2025-09-23", "2025-08-24"
,"2025-07-25"]



start_date = pd.Timestamp("2025-01-01")

# ACF for different dates (tarikh)
for xx in tarikh:

  end_date = pd.Timestamp(xx)

  # Inclusive bounds on both ends
  sliced_df[xx] = df[(df["Date"] >= start_date) & (df["Date"] <= end_date)].copy()
  sdf = sliced_df[xx]
  series = sdf["Close"].dropna()

  plt.figure(figsize=(10,5))
  plot_acf( series, lags=100, markersize=4)
  plt.title(f"ACF {xx}")

# PACF for different dates (tarikh)
for xx in tarikh:

  end_date = pd.Timestamp(xx)

  # Inclusive bounds on both ends
  sliced_df[xx] = df[(df["Date"] >= start_date) & (df["Date"] <= end_date)].copy()
  sdf = sliced_df[xx]
  series = sdf["Close"].dropna()

  plt.figure(figsize=(10,5))
  plot_pacf( series, lags=50, markersize=4)
  plt.title(f"PACF {xx}")




#print(list(sliced_df.keys()))
#print(list(sliced_df.values()))
```
